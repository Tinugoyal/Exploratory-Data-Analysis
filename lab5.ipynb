```python
import ipywidgets as widgets
from ipywidgets import interact
import matplotlib.pyplot as plt
```

```python

class FinancialGrowth:
    def __init__(self):

        self.__initial_deposit = 0.0
        self.__monthly_contribution = 0.0
        self.__annual_rate = 0.0
        self.__years = 0

    def set_parameters(self, initial: float, monthly: float, rate: float, years: int):
        self.__initial_deposit = float(initial)
        self.__monthly_contribution = float(monthly)
        self.__annual_rate = float(rate) / 100.0
        self.__years = int(years)

    def calculate_growth_curve(self):
        balances = []
        current_balance = self.__initial_deposit
        monthly_rate = self.__annual_rate / 12
        months = self.__years * 12

        for month in range(months + 1):
            if month % 12 == 0:
                balances.append(current_balance)

            interest = current_balance * monthly_rate
            current_balance += interest + self.__monthly_contribution

        return balances
```

```python
def plot_financial_growth(initial_deposit, monthly_savings, interest_rate, investment_years):

    tracker = FinancialGrowth()
    tracker.set_parameters(initial_deposit, monthly_savings, interest_rate, investment_years)
    yearly_balances = tracker.calculate_growth_curve()

    total_invested = initial_deposit + (monthly_savings * 12 * investment_years)
    final_wealth = yearly_balances[-1]
    total_interest_earned = final_wealth - total_invested

    print("=" * 50)
    print(f" Total Principal Invested:  ₹ {total_invested:,.2f}")
    print(f" Total Interest Compounded: ₹ {total_interest_earned:,.2f}")
    print(f" Final Portfolio Wealth:    ₹ {final_wealth:,.2f}")
    print("=" * 50)

    plt.figure(figsize=(10, 5))
    years_range = list(range(len(yearly_balances)))

    plt.plot(years_range, yearly_balances, marker='o', color='#2ca02c', linewidth=2, label="Portfolio Balance")
    plt.fill_between(years_range, yearly_balances, color='#2ca02c', alpha=0.15)

    plt.title("Compound Interest Wealth Accumulation Curve", fontsize=14, fontweight='bold', pad=15)
    plt.xlabel("Timeline (Years)", fontsize=11)
    plt.ylabel("Portfolio Value (₹)", fontsize=11)
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'₹{x:,.0f}'))
    plt.legend(loc="upper left")
    plt.show()

interact(
    plot_financial_growth,
    initial_deposit=widgets.FloatSlider(value=10000.0, min=0.0, max=500000.0, step=5000.0, description='Initial Principal:'),
    monthly_savings=widgets.FloatSlider(value=2000.0, min=0.0, max=50000.0, step=500.0, description='Monthly Save:'),
    interest_rate=widgets.FloatSlider(value=12.0, min=1.0, max=30.0, step=0.5, description='Return Rate %:'),
    investment_years=widgets.IntSlider(value=15, min=1, max=40, step=1, description='Years Plan:')
);
```
