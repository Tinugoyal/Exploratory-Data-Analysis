```python
import pandas as pd
```

Task 1 - Load the Dataset

```python
df = pd.read_csv('employee_salary.csv')
print("Dataset Shape:", df.shape)
print("No of rows: ",df.shape[0])
print("No of columns: ",df.shape[1])
print("\nFirst 5 Records:")
display(df.head())
print("\nLast 5 Records:")
display(df.tail())
```

Task 2 - Metadata and Statistics

```python

print("Column Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMetadata:")
df.info()

print("\nDescriptive Statistics:")
display(df.describe())
```

Task 3 - Checking Missing Values and Duplicates

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())

print("\nPercentage of Missing Values (%):")
print((df.isnull().sum() / len(df)) * 100)

print("Column with highest missing entries: ")
print(df.isnull().sum().idxmax())

print("Dataset Size Before Cleaning:", len(df))
print("Number of Duplicates:", df.duplicated().sum())

# Clean the dataset
df_cleaned = df.drop_duplicates()
print("Dataset Size After Cleaning:", len(df_cleaned))
```
